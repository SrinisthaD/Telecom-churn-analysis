```python
df=pd.read_csv('telecom_churn.csv')
```

```python
churn_by_calls=df.groupby("CustServCalls")['Churn'].mean()
```

```python
churn_by_calls
```

```python
bins=[0,1,3,5,100]
labels=['0','1-2','3-4','5+']
df['callsgroup']=pd.cut(df['CustServCalls'], bins=bins, labels=labels, right=False)
```

```python
churn=df.groupby('callsgroup')['Churn'].mean()*100
churn
```
